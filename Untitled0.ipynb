```python

!pip -q install flask requests

from flask import Flask, jsonify, request
from threading import Thread
import requests
import time

# Criando o servidor
app = Flask(__name__)

# Banco de dados simulado em memória
alunos = {}


# Rota inicial
@app.route("/", methods=["GET"])
def inicio():
    return jsonify({
        "mensagem": "Servidor Cloud funcionando!"
    })


# Cadastrar aluno
@app.route("/cadastrar", methods=["POST"])
def cadastrar():
    dados = request.get_json(silent=True)

    if not isinstance(dados, dict):
        return jsonify({"erro": "JSON invalido"}), 400

    nome = dados.get("nome")
    nota1 = dados.get("nota1")
    nota2 = dados.get("nota2")

    if not isinstance(nome, str) or not nome.strip():
        return jsonify({"erro": "Nome invalido"}), 400

    for nota in (nota1, nota2):
        if (
            isinstance(nota, bool)
            or not isinstance(nota, (int, float))
            or not 0 <= nota <= 10
        ):
            return jsonify({"erro": "Nota invalida"}), 400

    media = round((nota1 + nota2) / 2, 2)
    situacao = "Aprovado" if media >= 6 else "Reprovado"

    codigo = max(alunos.keys(), default=0) + 1

    alunos[codigo] = {
        "nome": nome.strip(),
        "nota1": nota1,
        "nota2": nota2,
        "media": media,
        "situacao": situacao
    }

    return jsonify({
        "mensagem": "Aluno cadastrado!",
        "codigo": codigo,
        "dados": alunos[codigo]
    }), 201


# Consultar todos os alunos
@app.route("/alunos", methods=["GET"])
def listar():
    return jsonify(alunos)


# Iniciar o servidor em segundo plano
def iniciar_servidor():
    app.run(
        host="127.0.0.1",
        port=5000,
        debug=False,
        use_reloader=False
    )


servidor = Thread(
    target=iniciar_servidor,
    daemon=True
)

servidor.start()
time.sleep(2)


# ==========================================
# CLIENTE ACESSANDO O SERVICO CLOUD
# ==========================================

url = "http://127.0.0.1:5000"

# Testar conexão
resposta = requests.get(url + "/", timeout=5)

print("STATUS DO SERVIDOR:")
print(resposta.json())


# Cadastrar alunos
dados_alunos = [
    {"nome": "Ana", "nota1": 8, "nota2": 9},
    {"nome": "Carlos", "nota1": 5, "nota2": 6},
    {"nome": "Mariana", "nota1": 10, "nota2": 9}
]

for aluno in dados_alunos:
    resposta = requests.post(
        url + "/cadastrar",
        json=aluno,
        timeout=5
    )

    print("\nCADASTRO:")
    print(resposta.status_code)
    print(resposta.json())


# Consultar alunos
resposta = requests.get(url + "/alunos", timeout=5)

print("\nALUNOS CADASTRADOS:")

for codigo, aluno in resposta.json().items():
    print(
        f"{codigo} - {aluno['nome']} | "
        f"Media: {aluno['media']} | "
        f"Situacao: {aluno['situacao']}"
    )
```
